```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = 'Diri.jpeg'
img_bgr = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
```

```python
img_cerah = cv2.convertScaleAbs(img_gray, alpha=1.0, beta=60)
img_kontras = cv2.convertScaleAbs(img_gray, alpha=1.5, beta=0)
img_cerah_kontras = cv2.convertScaleAbs(img_gray, alpha=1.5, beta=60)
```

```python
fig, axs = plt.subplots(2, 3, figsize=(15, 10))

axs[0, 0].imshow(img_rgb)
axs[0, 0].set_title('1. Citra Asli (RGB)')

axs[0, 1].imshow(img_gray, cmap='gray', vmin=0, vmax=255)
axs[0, 1].set_title('2. Citra Gray Asli')

axs[0, 2].imshow(img_cerah, cmap='gray', vmin=0, vmax=255)
axs[0, 2].set_title('3. Gray Dipercerah')

axs[1, 0].imshow(img_kontras, cmap='gray', vmin=0, vmax=255)
axs[1, 0].set_title('4. Gray Diperkontras')

axs[1, 1].imshow(img_cerah_kontras, cmap='gray', vmin=0, vmax=255)
axs[1, 1].set_title('5. Dipercerah & Diperkontras')

axs[1, 2].axis('off')

plt.tight_layout()
plt.show()
```
